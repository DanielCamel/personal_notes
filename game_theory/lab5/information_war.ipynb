```python
import numpy as np
import random

# Функция для вывода МАТРИЦЫ ДОВЕРИЯ
def print_matrix(matrix, title="Матрица доверия"):
    print(f"\n{title}:")
    for row in matrix:
        print(" ".join(f"{val:6.6f}" for val in row))

# Функция для вывода ВЕКТОРА МНЕНИЙ
def print_vector(vector, title="Вектор мнений"):
    print(f"\n{title}:")
    print("[ " + " ".join(f"{val:6.6f}" for val in vector) + " ]")

# Функция моделирования информационного противоборства
# X(t) = A * X(t-1) = A^2 * X(t-2) = ... = A^t * X(0)
def information_confrontation(X, matrix, error):
    E = 100  # Инициализация ошибки большим значением
    iteration = 1  # Счетчик итераций
    temp = np.zeros_like(X)  # Временный вектор для обновления мнений
    matrix_power = np.copy(matrix)  # Матрица для хранения A^t
    
    while E > error:  # Цикл выполняется, пока ошибка больше заданного порога
        min_val, max_val = float('inf'), float('-inf')
        for i in range(len(temp)):
            temp[i] = np.dot(matrix[i], X)  # Обновление м  нения агента через матрицу доверия
            min_val = min(min_val, temp[i])  # Поиск минимального значения
            max_val = max(max_val, temp[i])  # Поиск максимального значения
        X[:] = temp  # Обновление вектора мнений
        E = max_val - min_val  # Вычисление ошибки как разности максимума и минимума

        # Обновление степени матрицы A^t
        matrix_power = np.dot(matrix_power, matrix)  # Умножение матрицы на саму себя

        # Вектора мнений и степени матрицы доверия на каждой итерации
        # print(f"\nИтерация {iteration}:")
        # print_matrix(matrix_power, f"A^{iteration}")  # Выводим степень матрицы A^t
        # print_vector(X, "Вектор мнений")  # Выводим вектор мнений

        iteration += 1  # Увеличение счетчика итераций
    print(f"\nИтерация {iteration}:")
    print_matrix(matrix_power, f"A^{iteration}")  # Выводим степень матрицы A^t
    print_vector(X, "Вектор мнений")  # Выводим вектор мнений
    return iteration  # Возвращает количество итераций до сходимости

# Функция генерации агентов, принадлежащих двум противоборствующим группам
def gang_generation(N, X):
    a, b = random.randint(1, N//2 - 1), random.randint(1, N//2 - 1)
    gang_1 = random.sample(range(N), a)  # Генерация агентов первой группы
    gang_2 = list(set(random.sample(range(N), b)) - set(gang_1))  # Генерация агентов второй группы, исключая повторения
    
    gang_1_out = [i + 1 for i in gang_1]
    gang_2_out = [i + 1 for i in gang_2]

    print(f"\nАгенты первого игрока: {gang_1_out}")
    print(f"Агенты второго игрока: {gang_2_out}")
    
    force_1 = random.randint(0, 100)  # Формирование мнения первого игрока
    print(f"Начальное мнение первого игрока: U = {force_1}")
    for i in gang_1:
        X[i] = force_1
    
    force_2 = random.randint(-100, 0)  # Формирование мнения второго игрока
    print(f"Начальное мнение второго игрока: V = {force_2}")
    for i in gang_2:
        X[i] = force_2
    
    print_vector(X, "Начальные мнения после влияния")
    return force_1, force_2
```

```python
# Основной код программы
N = 10  # Количество агентов
print(f"Количество агентов: {N}")

# Генерация матрицы доверия 
matrix = np.random.rand(N, N)
# нормализация строк до суммы 1 по строкам
matrix /= matrix.sum(axis=1, keepdims=True)
print_matrix(matrix, "Матрица доверия")

# Инициализация начальных мнений агентов
X0 = np.random.rand(N) * 20
X = X0.copy()
print_vector(X, "Начальные мнения агентов")

error = 1e-6

print(f"\nПорог ошибки: {error}")
iterations = information_confrontation(X, matrix, error)
print(f"\nТребуемое количество итераций: {iterations}")
print_vector(X, "Конечные мнения (без влияния)")
print()
print("====================================================================")
print("====================================================================")
print()
# Применение влияния игроков
X = X0.copy()
U, V = gang_generation(N, X)
iterations = information_confrontation(X, matrix, error)
print(f"\nТребуемое количество итераций: {iterations}")
print_vector(X, "Конечные мнения (с влиянием)")
if abs(X[0] - U) >= 0:
    print("Выиграл 1-й игрок.")
else:
    print("Выиграл 2-й игрок.")    
```
