```python
import numpy as np
import random

# Функция генерации случайного числа в диапазоне
def f_rand(f_min, f_max):
    return random.uniform(f_min, f_max)

# Функция генерации случайного целого числа в диапазоне
def ran(i_min, i_max):
    return random.randint(i_min, i_max)

# Функция инициализации мнений агентов 
# Вектор мнений для НАЧАЛЬНОГО момента времени
def init_opinions(num_agents, lower_bound, upper_bound):
    return [f_rand(lower_bound, upper_bound) for i in range(num_agents)]

# Функция инициализации МАТРИЦЫ ДОВЕРИЯ
def fill_a_trust(num_agents):
    A_trust = []
    for i in range(num_agents):
        row = np.random.rand(num_agents)  # Генерация случайных значений
        row /= row.sum()  # Нормализация строк, чтобы сумма элементов была 1
        A_trust.append(row)
    return np.array(A_trust)

# # Функция вычисления нового мнения агента
# def sum_opinion(i, A, x):
#     return np.dot(A[i], x)



# Функция обновления мнений агентов
def change_opinion(A_trust, x_opinion):
    # Умножение матрицы доверия на вектор мнений
    new_opinion = np.matmul(A_trust, x_opinion)  
    return np.matmul(A_trust, A_trust), new_opinion

# Функция изменения мнений агентов под влиянием U и V
def new_u_v(A_trust, x_opinion, players_U, players_V, players_n, init_op):
    U = ran(30, 90)
    V = -U
    print(f"U = {U} V = {V}")
    
    for pl in players_U:
        x_opinion[pl] = np.sum(A_trust[pl]) * U
    
    for pl in players_V:
        x_opinion[pl] = np.sum(A_trust[pl]) * V
    
    for pl in players_n:
        x_opinion[pl] = np.dot(A_trust[pl], init_op)
    
    print("Агенты под влиянием U:", ' '.join(map(str, [p + 1 for p in players_U])))
    print("Агенты под влиянием V:", ' '.join(map(str, [p + 1 for p in players_V])))
    
    return x_opinion

# Функция проверки сходимости матрицы доверия
def equal_a_trust(A_trust, err):
    diff = np.abs(A_trust - np.mean(A_trust, axis=1, keepdims=True))
    return np.all(diff < err)

# Функция проверки сходимости мнений агентов
def equal_opinion(x_opinion, err):
    diff = np.abs(x_opinion - np.mean(x_opinion))
    return np.all(diff < err)

# Функция вывода ТЕКУЩЕГО состояния системы
def print_state(A_trust, x_opinion):
    print("\nМатрица доверия =")
    for row in A_trust:
        print("|", " ".join(f"{x:12.6f}" for x in row), "|")
    print()
    print("Вектор мнений:", " ".join(f"{x:12.6f}" for x in x_opinion))

```

```python
# Основной процесс моделирования

# Количество агентов
number_agents = 10
# Границы значений мнений агентов 
# нижняя
lower_bound = 1
# верхняя
upper_bound = 20

# Порог ошибки для сходимости
ERR = 1e-2

x_opinion = init_opinions(number_agents, lower_bound, upper_bound)
A_trust = fill_a_trust(number_agents)
init_matrix = A_trust.copy()
init_op = x_opinion.copy()

players_U = [0, 1, 2]
players_V = [3, 4, 5]
players_n = [6, 7, 8, 9]

print_state(A_trust, x_opinion)

for n in range(1, 10):
    A_trust, x_opinion = change_opinion(A_trust, x_opinion)
    if equal_a_trust(A_trust, ERR) and equal_opinion(x_opinion, ERR):
        print(f"\n\nN = {n}   FINAL:")
        print_state(A_trust, x_opinion)
        break

x_opinion = init_op.copy()
x_opinion = new_u_v(A_trust, x_opinion, players_U, players_V, players_n, init_op)
print_state(A_trust, x_opinion)

for n in range(1, 100):
    A_trust, x_opinion = change_opinion(A_trust, x_opinion)
    if equal_a_trust(A_trust, ERR) and equal_opinion(x_opinion, ERR):
        print(f"\n\nN = {n}   FINAL:")
        print_state(A_trust, x_opinion)
        break
```
