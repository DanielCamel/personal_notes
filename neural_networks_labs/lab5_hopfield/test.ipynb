```python
import numpy as np
import random
def Print_Array(vector,vector_n=5,vector_m=3):
    for index_i in range(vector_n):
        for index_j in range(vector_m):
            if vector[index_i*array_m+index_j]==1:
                print("*", end = "")
            else:
                print(" ", end = "")
        print()
    
def Pomeha_Array(vector,vector_n=5,vector_m=3):
    for index_i in range(vector_n):
        for index_j in range(vector_m):
            if random.randint(0, 9)%4==1:
                if vector[index_i*array_m+index_j]==1:
                    vector[index_i*array_m+index_j]=-1
                else:
                    vector[index_i*array_m+index_j]=1

def function_activate(net, pred):
    if (net>0):
        return 1
    elif (net<0):
        return -1
    else:
        return pred

def learning_verification(current_signal, previous_signal):
    for index_i in range(len(current_signal)):
        if current_signal[index_i]!=previous_signal[index_i]:
            return 0
    return 1

def Raschet_net(matrix_weight,signal):
    signal_size=len(matrix_weight[0])
    current_signal = [signal[i] for i in range(signal_size)]
    previous_signal = [-1 for i in range(signal_size)]
    era = 0
    network_output = [0 for i in range(signal_size)]
    while not learning_verification(current_signal, previous_signal):
       # print("Текущий сигнал: "+str(current_signal))
        for index_i in range(signal_size):
            network_output[index_i]=0
            previous_signal[index_i]=current_signal[index_i]
        for index_k in range(signal_size):
            for index_j in range(signal_size):
            #     network_output[index_k]+=matrix_weight[index_k][index_j]*current_signal[index_j]
            # for index_j in range(index_k+1,signal_size):
                network_output[index_k]+=matrix_weight[index_k][index_j] * previous_signal[index_j]
            current_signal[index_k] = function_activate(network_output[index_k],previous_signal[index_k])
        era+=1
        #print("Эпоха обучения: "+str(era))

    return current_signal
```
