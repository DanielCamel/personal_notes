```python
import numpy as np

class HopfieldNetwork:
    def __init__(self, size):
        """Создает сеть Хопфилда с матрицей весов size x size."""
        self.size = size
        self.weights = np.zeros((size, size))

    def train(self, patterns):
        """Обучает сеть по правилу Хебба."""
        self.weights.fill(0)
        for p in patterns:
            p = np.array(p).reshape(-1, 1)
            self.weights += np.dot(p, p.T)
        np.fill_diagonal(self.weights, 0)

    def activate(self, x):
        """Активационная функция."""
        return np.where(x >= 0, 1, -1)

    def recall(self, pattern, steps=10):
        """Восстановление образа."""
        pattern = np.array(pattern)
        for _ in range(steps):
            pattern = self.activate(np.dot(self.weights, pattern))
        return pattern

def print_pattern(pattern, rows, cols, title="Образ"):
    """Вывод образа в виде *"""
    print("\n" + title + "\n")
    pattern_reshaped = pattern.reshape(rows, cols)
    for row in pattern_reshaped:
        print("".join("*" if pixel == 1 else " " for pixel in row))
    print()

# === Размерность образов ===
rows, cols = 25, 20  # 25x20 звездочек

# === Функции генерации "жирных" букв ===
def create_X(rows, cols):
    """Создает букву X с утолщенными линиями."""
    X = np.full((rows, cols), -1)
    for i in range(rows):
        for j in range(cols):
            if abs(i - j) < 2 or abs(i + j - (cols - 1)) < 2:  # Утолщаем диагонали
                X[i, j] = 1
    return X.flatten()

def create_Y(rows, cols):
    """Создает букву Y с утолщенными линиями."""
    Y = np.full((rows, cols), -1)
    mid = cols // 2
    for i in range(rows):
        for j in range(cols):
            if (i < rows // 2 and (abs(i - j) < 2 or abs(i + j - (cols - 1)) < 2)) or (i >= rows // 2 and abs(j - mid) < 2):
                Y[i, j] = 1
    return Y.flatten()

def create_Z(rows, cols):
    """Создает букву Z с утолщенными линиями."""
    Z = np.full((rows, cols), -1)
    for i in range(rows):
        for j in range(cols):
            if i < 2 or i > rows - 3 or abs(j - ((rows - i - 1) * cols // rows)) < 2:  # Утолщенные верхняя, нижняя и диагональная линии
                Z[i, j] = 1
    return Z.flatten()
```

```python
# === Размерность образов ===
rows, cols = 15, 10  # 15x10 звездочек

# === Функции генерации "жирных" букв ===
def create_X(rows, cols):
    """Создает букву X"""
    X = np.full((rows, cols), -1)
    for i in range(rows):
        for j in range(cols):
            if abs(i - j) < 2 or abs(i + j - (cols - 1)) < 2:  # Утолщаем диагонали
                X[i, j] = 1
    return X.flatten()

def create_Y(rows, cols):
    """Создает букву Y"""
    Y = np.full((rows, cols), -1)
    mid = cols // 2
    for i in range(rows):
        for j in range(cols):
            if (i < rows // 2 and (abs(i - j) < 2 or abs(i + j - (cols - 1)) < 2)) or (i >= rows // 2 and abs(j - mid) < 2):
                Y[i, j] = 1
    return Y.flatten()

def create_Z(rows, cols):
    """Создает букву Z"""
    Z = np.full((rows, cols), -1)
    for i in range(rows):
        for j in range(cols):
            if i < 2 or i > rows - 3 or abs(j - ((rows - i - 1) * cols // rows)) < 2:  # Утолщенные верхняя, нижняя и диагональная линии
                Z[i, j] = 1
    return Z.flatten()

# Создаем утолщенные образы X, Y, Z
X = create_X(rows, cols)
Y = create_Y(rows, cols)
Z = create_Z(rows, cols)

# Создаем сеть Хопфилда и обучаем её
hopfield_net = HopfieldNetwork(size=rows * cols)
hopfield_net.train([X, Y, Z])

# === Вывод эталонных образов ===
print("\nЭталонные образы:")
print_pattern(X, rows, cols, title="Образ X")
print_pattern(Y, rows, cols, title="Образ Y")
print_pattern(Z, rows, cols, title="Образ Z")

# === Восстановление ===
print("\nТестирование на эталонных образах:")
print_pattern(hopfield_net.recall(X), rows, cols, title="Восстановленный X")
print_pattern(hopfield_net.recall(Y), rows, cols, title="Восстановленный Y")
print_pattern(hopfield_net.recall(Z), rows, cols, title="Восстановленный Z")

# === Создаем искаженный X ===
distorted_X = np.copy(X)
distorted_X[5 * cols + 2] *= -1  # Инверсия пикселей
distorted_X[7 * cols + 3] *= -1
distorted_X[9 * cols + 4] *= -1

print("\nТестирование на искаженном X:")
print_pattern(distorted_X, rows, cols, title="Образ X с помехой")

# print("\nТестирование на искаженном Y:")
# print_pattern(distorted_Y, rows, cols, title="Образ Y с помехой")

# print("\nТестирование на искаженном Z:")
# print_pattern(distorted_Z, rows, cols, title="Образ Z с помехой")

restored_X = hopfield_net.recall(distorted_X)
print_pattern(restored_X, rows, cols, title="Восстановленный X после помехи")
```
