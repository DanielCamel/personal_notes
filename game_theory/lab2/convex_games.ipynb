```python
import numpy as np

def H(x, y):
    return -6*(x**2)+16/5*(y**2)+16*x*y-16/5*x-48/5*y

# Метод Брауна–Робинсон (итерационный алгоритм)
def braun_robinson(C, epsilon):
    n, m = C.shape # число стратегий игроков

    A = np.zeros(n) # массив из n нулей
    B = np.zeros(m) # массив из m нулей

    x_count = np.zeros(n)
    y_count = np.zeros(m)

    A = C[:,0] # первый столбец C
    B = C[0,:] # первая строка C

    x_count[0] += 1
    y_count[0] += 1
    
    I_A = 0
    I_B = 0

    v_up_arr = []
    v_down_arr= []

    eps = 1000
    k = 0

    while eps >= epsilon:
        k+=1 # номер итерации
        x_count[I_A] += 1 # увеличение числа выборов одной из стратегий игрока А
        y_count[I_B] += 1 # увеличение числа выборов одной из стратегий игрока B
        v_up   = max(A)/k # верхняя цена игры
        v_down = min(B)/k # нижняя цена игры
        v_up_arr.append(v_up) # массив верхних цен
        v_down_arr.append(v_down) # массив нижних цен

        eps = min(v_up_arr) - max(v_down_arr) # погрешность  
        I_A = np.argmax(A) # индекс максимального элемента стратегии 
        I_B = np.argmin(B) # индекс минимального элемента стратегии
        A = A+C[:,I_B]
        B = B+C[I_A,:]
    return x_count/k, y_count/k # приближенные смешанные стратегии
    
# основная программа


for N in range(1,11):
    print("N=",N,"\n")
    matr = np.zeros((N+1,N+1))
    for i in range(N+1):
        for j in range(N+1):
            x = i/N
            y = j/N
            matr[i][j] = round(H(x, y),3)

    print(matr,"\n")
    # максимальные элементы в столбцах
    col_max = np.argmax(matr,axis=0) # [1 0 2]
    # минимальные элементы в строках 
    raw_min = np.argmin(matr,axis=1)

    arr_max_elem = []
    arr_min_elem = []
    n, m = matr.shape
    # массив минимальных элементов по строкам
    for p in range(n):
        arr_min_elem.append(matr[p,raw_min[p]]) 
    # массив максимальных элементов по столбцам    
    for l in range(m):
        arr_max_elem.append(matr[col_max[l],l])
    v_down_N = max(arr_min_elem) # Нижняя цена игры
    v_up_N = min(arr_max_elem)   # Верхняя цена игры

    if v_down_N == v_up_N:
        H_point = v_up_N
        x_point = round(np.argmax(arr_min_elem)/N,3)
        y_point = round(np.argmin(arr_max_elem)/N,3)
        print("Есть седловая точка:\n")
        print("x=",x_point,"y=",y_point," ","H=",H_point)
    else:
        x_opt, y_opt = braun_robinson(matr,epsilon=0.01)
        x_point = np.argmax(x_opt)
        y_point = np.argmax(y_opt)
        print("Седловой точки нет, решение методом Брауна-Робинсон:\n")
        print("x=",x_point/N,"y=",y_point/N," ","H=",matr[x_point][y_point],"\n")   




```
