```python
import numpy as np
import random


def recognition(current_signal, vector1, vector2, vector3, show = False):

        if np.all(current_signal == vector1):
            if show:
                print("This is X image\n")
            return vector1
        elif np.all(current_signal == vector2):
            if show:
                print("This is Y image\n")
            return vector2
        elif np.all(current_signal == vector3):
            if show:
                print("This is Z image\n")
            return vector3
        else:
            print("Error! Object is not defined!\n")
            return 0


def print_pattern(vector,rows=15,cols=10):
    for i in range(rows):
        for j in range(cols):
            if vector[(i * cols) + j] == 1:
                print("*", end = "")
            else:
                print(" ", end = "")
        print()

def activation_function(net, f_pred):
    if (net > 0):
        return 1
    elif (net < 0):
        return -1
    else:
        return f_pred

def learning_verification(current_signal, previous_signal):
    for i in range(len(current_signal)):
        if current_signal[i] != previous_signal[i]:
            return 0
    return 1

def calculation_net(matrix_weight, pattern, max_epochs = 100):
    signal_size = len(matrix_weight[0])
    
    # Текущий сигнал (копия входного вектора)
    current_signal = np.array(pattern, dtype=int)

    # Предыдущий сигнал (чтобы отслеживать изменения)
    previous_signal = np.full(signal_size, -1, dtype=int)

    # Вектор сетевых выходов
    network_output = np.zeros(signal_size, dtype=int)

    epoch = 0

    while not learning_verification(current_signal, previous_signal) and epoch < max_epochs:
        # print(f"\nЭпоха {epoch + 1}:")
        
        for i in range(signal_size):
            network_output[i] = 0
            previous_signal[i] = current_signal[i]
        
        # Асинхронный режим
        # for k in range(signal_size):
        #     for j in range(k):
        #         network_output[k]+=matrix_weight[k][j]*current_signal[j]
        #     for j in range(k+1,signal_size):
        #         network_output[k]+=matrix_weight[k][j]*previous_signal[j]
            # current_signal[k] = activation_function(network_output[k], previous_signal[i])
        # Синхронный режим
        for k in range(signal_size):
            for j in range(signal_size):
                network_output[k] += matrix_weight[k][j] * previous_signal[j]
            
            # Применяем функцию активации
            current_signal[k] = activation_function(network_output[k], previous_signal[k])
        epoch += 1

    return current_signal

def disturbance_pattern(pattern, rows = 15, cols=10, percentage = 30):
    """Искажает примерно percentage% элементов во входном векторе pattern."""
    distort_pattern = np.full(len(pattern), -1)
    for j in range(len(pattern)):
        distort_pattern[j] = int(pattern[j])
    total_elements = rows * cols
    changes = total_elements * percentage // 100
    indices = random.sample(range(total_elements), changes)  # Выбираем случайные индексы
    for idx in indices:
        distort_pattern[idx] *= -1  # Инвертируем значение
    return distort_pattern  # Возвращаем измененный паттерн
```
