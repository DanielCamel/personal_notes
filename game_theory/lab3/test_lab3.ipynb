```python
import numpy as np

class BiMatrix:
    def __init__(self, rows, columns, game_type=None):
        self.row = rows
        self.col = columns
        self.game_type = game_type

        if game_type == "ordinar":
            self.matrix_A = np.random.randint(-99, 99, size=(rows, columns))
            self.matrix_B = np.random.randint(-99, 99, size=(rows, columns))
        elif game_type == "crossroad":
            self.matrix_A = np.array([[1, 0.5], [2, 0]])
            self.matrix_B = np.array([[1, 2], [0.5, 0]])
        elif game_type == "prisoner":
            self.matrix_A = np.array([[-5, 0], [-10, -1]])
            self.matrix_B = np.array([[-5, -10], [0, -1]])
        elif game_type == "family":
            self.matrix_A = np.array([[4, 0], [0, 1]])
            self.matrix_B = np.array([[1, 0], [0, 4]])
        else:
            self.matrix_A = np.zeros((rows, columns))
            self.matrix_B = np.zeros((rows, columns))

    def my_bimatrix(self, a, b):
        self.matrix_A = a
        self.matrix_B = b

    def find_nash(self, a, b):
        rows, cols = a.shape
        nash_equilibria = []
        for i in range(rows):
            for j in range(cols):
                payoff_a, payoff_b = a[i, j], b[i, j]
                best_response_a = all(payoff_a >= a[ii, j] for ii in range(rows))
                best_response_b = all(payoff_b >= b[i, jj] for jj in range(cols))
                if best_response_a and best_response_b:
                    nash_equilibria.append((i, j))
        return nash_equilibria

    def find_pareto(self, a, b):
        rows, cols = a.shape
        all_pairs = [(i, j, a[i, j], b[i, j]) for i in range(rows) for j in range(cols)]
        pareto_optimal = []
        for (i1, j1, ai, bi) in all_pairs:
            is_pareto_optimal = True
            for (i2, j2, ak, bk) in all_pairs:
                if (ak >= ai and bk > bi) or (ak > ai and bk >= bi):
                    is_pareto_optimal = False
                    break
            if is_pareto_optimal:
                pareto_optimal.append((i1, j1))
        return pareto_optimal

    def bimatrix_output(self, a, b):
        nash_eq = self.find_nash(a, b)
        pareto_opt = self.find_pareto(a, b)
        
        print("Биматричная игра:")
        if self.game_type:
            print(f"Тип игры: {self.game_type.capitalize()}")
        
        for i in range(self.row):
            row_str = []
            for j in range(self.col):
                cell = f"({a[i, j]:>4}, {b[i, j]:>4})"
                if (i, j) in nash_eq:
                    cell = f"\033[34m{cell}\033[0m"  # Синий цвет
                if (i, j) in pareto_opt:
                    cell = f"\033[31m{cell}\033[0m"  # Красный цвет
                row_str.append(cell)
            print("  ".join(row_str))

    def generate_game(self):
        a, b = self.matrix_A, self.matrix_B
        self.bimatrix_output(a, b)

obj = BiMatrix(10, 10, "ordinar")
obj.generate_game()
```
